```python
import pandas as pd
print(pd.__version__)
```

```python
import numpy as np

url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

```python
df["fuel_type"].nunique()
```

```python
missing_per_column = df.isna().sum()
print(missing_per_column)
```

```python
asia_cars = df[df["origin"] == "Asia"]
asia_cars["fuel_efficiency_mpg"].max()
```

```python
median_before = df["horsepower"].median()
most_frequent = df["horsepower"].mode().iloc[0]
median_after = df["horsepower"].fillna(most_frequent).median()

print("Median before:", median_before)
print("Most frequent value:", most_frequent)
print("Median after:", median_after)
```

```python
X = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]].head(7).to_numpy()

print("Shape:", X.shape)
print(X)
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print("Weights:", w)
print("Sum:", w.sum())
```
